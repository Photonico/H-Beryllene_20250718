# Dielectic function and linear optical properties for thesis

Optical normalization uses the relaxed structure of each calculation and the original surface-radius convention (Be: 1.98 Å; H: 0.70 Å). Hydrogenated slabs use the species at the outermost atomic centres; this convention is not a union of atomic spheres. Both xx and zz retain the same effective-volume normalization. Absorption is reported in nm$^{-1}$ with $\omega=2\pi E/h$. These are effective-thickness IPA interband spectra; no Drude term is added. Scalar reflectivity uses the semi-infinite-medium expression, not a finite-sheet measurement model.

```python
# prework

from vmatplot.linear_optical_properties import plot_linear_optical_property
from vmatplot.dielectric_function import plot_dielectric_function
from vmatplot.structure import slab_optical_factor
from vmatplot.plot_validation import ensure_curve_visibility
import matplotlib.pyplot as plt

components = [{"xx":"in-plane"}, {"zz":"out-of-plane"}]
photon_boundary = (0, 12)
data_boundary = (-60, 80)
layout = "hor"

# Optical factors from each relaxed structure; radii: Be = 1.98 A, H = 0.70 A.
# The cell height is the full surface-normal period, not the vacuum gap.
factor_a = slab_optical_factor("6.0_dielectric_selection/a-Beryllene")
factor_b = slab_optical_factor("6.0_dielectric_selection/b-Beryllene")

# systems list
systems = [["α-beryllene",              "6.0_dielectric_selection/a-Beryllene",  "blue",     None,None,None, factor_a],
           ["β-beryllene",              "6.0_dielectric_selection/b-Beryllene",  "orange",   None,None,None, factor_b]]
```

```python
# dielectric function

plot_dielectric_function("Dielectric function", systems, components, "hor", True, "eV", (0,12), data_boundary, True)
ensure_curve_visibility(plt.gcf())
plt.savefig("figures_for_publication/fig3.13_dielectric.pdf")

fig = plt.gcf()
# 1. suptitle fontsize
if fig._suptitle is not None:
    fig._suptitle.set_fontsize(28)
# all subplots
for ax in fig.get_axes():
    # 2. subtitle fontsize
    ax.title.set_fontsize(24)
    # 3. axes label fontsize
    ax.xaxis.label.set_fontsize(20)
    ax.yaxis.label.set_fontsize(20)
    # 4. axes ticks fontsize
    ax.tick_params(axis="both", which="major", labelsize=16)
    ax.tick_params(axis="both", which="minor", labelsize=16)
    # 5. Legend fontsize
    legend = ax.get_legend()
    if legend is not None:
        for text in legend.get_texts():
            text.set_fontsize(18)
        legend.get_title().set_fontsize(18)
plt.tight_layout()

ensure_curve_visibility(plt.gcf())
plt.savefig("figures_for_thesis/3.1_dielectric.pdf")
```

```python
# Absorption coefficient

value_boundary = (-0.0125, 0.2625)
plot_linear_optical_property("Absorption coefficient", systems, "absorption", components, layout, True, "eV", photon_boundary, value_boundary, True)
ensure_curve_visibility(plt.gcf())
plt.savefig("figures_for_publication/fig3.14a_absorption.pdf")

fig = plt.gcf()
# 1. suptitle fontsize
if fig._suptitle is not None:
    fig._suptitle.set_fontsize(28)
# all subplots
for ax in fig.get_axes():
    # 2. subtitle fontsize
    ax.title.set_fontsize(24)
    # 3. axes label fontsize
    ax.xaxis.label.set_fontsize(20)
    ax.yaxis.label.set_fontsize(20)
    # 4. axes ticks fontsize
    ax.tick_params(axis="both", which="major", labelsize=16)
    ax.tick_params(axis="both", which="minor", labelsize=16)
    # 5. Legend fontsize
    legend = ax.get_legend()
    if legend is not None:
        for text in legend.get_texts():
            text.set_fontsize(18)
        legend.get_title().set_fontsize(18)
plt.tight_layout()

ensure_curve_visibility(plt.gcf())
plt.savefig("figures_for_thesis/4.1_absorption.pdf")
```

# Band structure

```python
# prework

from vmatplot.bandstructure import plot_bandstructure
import matplotlib.pyplot as plt
```

```python
# α-Beryllene_hh

bs_list = [["monocolor", "α-Beryllene", "4.0_bandstructure/a-Beryllene_hh", "blue"]]

plot_bandstructure("α-Beryllene with Hydrogen (double side adhesion)", bs_list, (-8,8))
plt.savefig("figures/4_bs/4.31_BS_a_hh.pdf")
```

```python
# β-Beryllene_h

bs_list = [["monocolor", "β-Beryllene", "4.0_bandstructure/b-Beryllene_h", "blue"]]

plot_bandstructure("β-Beryllene with Hydrogen (single side adhesion)", bs_list, (-8,8))
plt.savefig("figures/4_bs/4.32_BS_b_h.pdf")
```

```python
# β-Beryllene_hh

bs_list = [["monocolor", "β-Beryllene", "4.0_bandstructure/b-Beryllene_hh", "blue"]]

plot_bandstructure("β-Beryllene with Hydrogen (double side adhesion)", bs_list, (-8,8))
plt.savefig("figures/4_bs/4.33_BS_b_hh.pdf")
```

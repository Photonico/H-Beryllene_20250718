# Band structure

```python
# prework

from vmatplot.bandstructure import plot_bandstructure
import matplotlib.pyplot as plt
from vmatplot.output_settings import figure_version, save_figure

figure_version("article")    # "thesis": thesis typography, figures saved as figures/<name>_thesis.pdf

# Zero of energy: the Fermi energy of the self-consistent run (last entry of each bs_list), not that of the
# non-self-consistent band run, whose E-fermi is obtained from the path k-points only
```

```python
# α-beryllene_hh: PBE (4.0_bandstructure, blue) and HSE06 (4.4_bandstructure_hse, orange), both aligned at the VBM

from vmatplot.band_topology import plot_hse_bands

plot_hse_bands("2H-α-beryllene",
               ["α-beryllene", "4.4_bandstructure_hse/a-Beryllene_hh", "4.0_bandstructure/a-Beryllene_hh", "Orange", "Blue"], (-8,8))
save_figure("bands_2H-alpha-beryllene")
# fig = plt.gcf()
# fig.set_size_inches(8, 6)
```

```python
# β-beryllene_b: path Γ–X–M–Γ–Y–M′–Γ through all four TRIM, as for the phonons

bs_list = [["monocolor", "β-beryllene", "4.0_bandstructure/b-Beryllene_b_trim", "blue", "solid", 1.5, 1.0, 0,
            "2.5_self_consistent_with_hydrogen/b-Beryllene_h"]]

plot_bandstructure("1H-β-beryllene", bs_list, (-8,8))
fig = plt.gcf()
fig.set_size_inches(9, 6)
save_figure("bands_1H-beta-beryllene")
```

```python
# β-beryllene_bb: path Γ–X–M–Γ–Y–M′–Γ through all four TRIM, as for the phonons

bs_list = [["monocolor", "β-beryllene", "4.0_bandstructure/b-Beryllene_bb_trim", "blue", "solid", 1.5, 1.0, 0,
            "2.5_self_consistent_with_hydrogen/b-Beryllene_hh"]]

plot_bandstructure("2H-β-beryllene", bs_list, (-8,8))
fig = plt.gcf()
fig.set_size_inches(9, 6)
save_figure("bands_2H-beta-beryllene")
```

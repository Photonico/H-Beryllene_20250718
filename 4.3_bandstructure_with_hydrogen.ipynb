# Band structure

```python
# prework

from vmatplot.bandstructure import plot_bandstructure
import matplotlib.pyplot as plt
```

```python
# α-beryllene_hh

bs_list = [["monocolor", "α-beryllene", "4.0_bandstructure/a-beryllene_hh", "blue"]]

plot_bandstructure("double-sided H-terminated α-beryllene", bs_list, (-8,8))
plt.savefig("figures/4_bs/4.31_BS_a_hh.pdf")
plt.savefig("figures_for_publication/fig3.9_bs_a-beryllene_hh.pdf")
# fig = plt.gcf()
# fig.set_size_inches(8, 6)
# plt.savefig("figures_for_publication/fig3.9_bs_a-beryllene_hh_alt.pdf")
```

```python
# β-beryllene_b

bs_list = [["monocolor", "β-beryllene", "4.0_bandstructure/b-beryllene_b_ollie", "blue"]]

plot_bandstructure("H-terminated β-beryllene", bs_list, (-8,8))
plt.savefig("figures/4_bs/4.32_BS_b_b.pdf")
plt.savefig("figures_for_publication/fig3.10a_bs_b-beryllene_b.pdf")
fig = plt.gcf()
fig.set_size_inches(9, 6)
plt.savefig("figures_for_publication/fig3.10a_bs_b-beryllene_b_alt.pdf")
```

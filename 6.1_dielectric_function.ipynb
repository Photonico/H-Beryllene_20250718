# Dielectric function


```python
# Prework

from vmatplot.dielectric_function import *
import matplotlib.pyplot as plt

# system factors (d_c, (z_max-z_min)*d_c+2*r_be) r_be = 1.98
factor_bulk = (1, 1)                                        # for α-Beryllium and cubic Beryllium bulk
factor_a = (40, 2*1.98)                                     # for α-Beryllene monolayer
factor_b = (40, 0.0473771641975619+2*1.98)                  # for β-Beryllene monolayer
factor_c = (40, 0.0751622445053417+2*1.98)                  # for cubic Beryllene trilayer

data_boundary = (-60, 60)
```

```python
# dielectric function

systems = [
           ["α-Beryllium bulk",         "6.0_dielectric_selection/a-Beryllium",  "violet",   None,None,None, factor_bulk],
           ["α-Beryllene",              "6.0_dielectric_selection/a-Beryllene",  "blue",     None,None,None, factor_a],
           ["β-Beryllene",              "6.0_dielectric_selection/b-Beryllene",  "lime",     None,None,None, factor_b],
           ["cubic Beryllium bulk",     "6.0_dielectric_selection/c0-Beryllium", "purple",   None,None,None, factor_bulk],
           ["cubic Beryllene trilayer", "6.0_dielectric_selection/c3-Beryllene", "orange",   None,None,None, factor_c]
           ]

components = [{"xx":"xx-component"},{"yy":"yy-component"},{"zz":"zz-component"}]

plot_dielectric_function("Dielectric function", systems, components, "hor", True, "eV", (0,12), data_boundary, True)
plt.savefig("figures/6_dielectric/6.30_dielec.pdf")
```

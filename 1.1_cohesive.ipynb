# Cohesive energy testing

Cohesive energy versus parameters


```python
# prework

from vmatplot.convergence import summarize_cohesive_energy, plot_cohesive_energy_parameters
import matplotlib.pyplot as plt

encut_boundary = (300, None)
kpoints_boundary = (12**2, None)
```

```python
# cohesive energy versus energy cutoff

summarize_cohesive_energy("1.0_convergence/g-Beryllene_energycutoff","1.0_convergence/single_Beryllium")
g_encut = ["g-Beryllene", "1.0_convergence/g-Beryllene_energycutoff/cohesive_energy.dat", encut_boundary, "blue"]
encut_list = [g_encut]

plot_cohesive_energy_parameters("Cohesive energy versus energy cutoff for g-Beryllene", "encut", encut_list )
plt.savefig("figures/1.10_cohesive_encut.pdf")
```

```python
# cohesive energy versus k-points for a-Beryllene

summarize_cohesive_energy("1.0_convergence/a-Beryllene_kpoints","1.0_convergence/single_Beryllium")
a_kpoints = ["a-Beryllene", "1.0_convergence/a-Beryllene_kpoints/cohesive_energy.dat", kpoints_boundary, "violet"]
kpoints_list = [a_kpoints]

plot_cohesive_energy_parameters("Cohesive energy versus k-points for a-Beryllene", "kpoints", kpoints_list)
plt.savefig("figures/1.11_kpoints_a.pdf")
```

```python
# cohesive energy versus k-points for g-Beryllene

summarize_cohesive_energy("1.0_convergence/g-Beryllene_kpoints","1.0_convergence/single_Beryllium")
g_kpoints = ["g-Beryllene", "1.0_convergence/g-Beryllene_kpoints/cohesive_energy.dat", kpoints_boundary, "blue"]
kpoints_list = [g_kpoints]

plot_cohesive_energy_parameters("Cohesive energy versus k-points for g-Beryllene", "kpoints", kpoints_list)
plt.savefig("figures/1.12_kpoints_g.pdf")
```

```python
# cohesive energy versus k-points for c-Beryllene

summarize_cohesive_energy("1.0_convergence/c-Beryllene_kpoints","1.0_convergence/single_Beryllium")
c_kpoints = ["c-Beryllene", "1.0_convergence/c-Beryllene_kpoints/cohesive_energy.dat", kpoints_boundary, "lime"]
kpoints_list = [c_kpoints]

plot_cohesive_energy_parameters("Cohesive energy versus k-points for c-Beryllene", "kpoints", kpoints_list)
plt.savefig("figures/1.13_kpoints_c.pdf")
```

```python
# cohesive energy versus k-points for c-Beryllene trilayer

summarize_cohesive_energy("1.0_convergence/c-Beryllene_trilayer_kpoints","1.0_convergence/single_Beryllium")
t_kpoints = ["c-Beryllene trilayer", "1.0_convergence/c-Beryllene_trilayer_kpoints/cohesive_energy.dat", kpoints_boundary, "yellow"]
kpoints_list = [t_kpoints]

plot_cohesive_energy_parameters("Cohesive energy versus k-points for c-Beryllene trilayer", "kpoints", kpoints_list)
plt.savefig("figures/1.14_kpoints_t.pdf")
```

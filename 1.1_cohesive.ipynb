# Cohesive energy testing

Cohesive energy versus parameters


```python
# prework

from vmatplot.convergence import summarize_cohesive_energy, plot_cohesive_energy_encut, plot_cohesive_energy_parameters
import matplotlib.pyplot as plt

encut_boundary = (300, None)
kpoints_boundary = (18**2, None)
```

```python
# cohesive energy versus energy cutoff

summarize_cohesive_energy("1.0_convergence/g-Beryllene_energycutoff","1.0_convergence/Atom_Beryllium")
g_encut = ["g-Beryllene", "1.0_convergence/g-Beryllene_energycutoff/cohesive_energy.dat", encut_boundary, "blue"]

plot_cohesive_energy_parameters("Cohesive energy versus energy cutoff for g-Beryllene", "encut", [g_encut])
plt.savefig("figures/1.1a_cohesive_encut.pdf")
```

```python
# cohesive energy versus k-points

summarize_cohesive_energy("1.0_convergence/g-Beryllene_kpoints","1.0_convergence/Atom_Beryllium")
g_kpoints = ["g-Beryllene", "1.0_convergence/g-Beryllene_kpoints/cohesive_energy.dat", kpoints_boundary, "blue"]

plot_cohesive_energy_parameters("Cohesive energy versus k-points mesh for g-Beryllene", "kpoints", [g_kpoints])
plt.savefig("figures/1.1b_cohesive_kpoints.pdf")
```

# Geometry information with Hydrogen adhesion

```python
# prework

from vmatplot.commons import identify_parameters
import matplotlib.pyplot as plt
```

```python
# a-Beryllene hollow-hollow

geometry_data = identify_parameters("2.5_self_consistent_with_hydrogen/a-Beryllene_hh")
lattice = geometry_data["lattice constant"]
kpoints = geometry_data["kpoints mesh"]
energy = geometry_data["total energy"]
time = geometry_data["elapsed time (sec)"]

print(f"For α-Beryllene double side adhesion: \n k-points mesh: {kpoints}; \n lattice constant: {lattice}; \n total energy: {energy}; \n elapsed time: {time}.")
```

# Cohesive energy testing for cubic family

Cohesive energy versus parameters


```python
# prework

from vmatplot.convergence import summarize_cohesive_energy, plot_cohesive_energy_parameters
import matplotlib.pyplot as plt

encut_boundary = (300, None)
kpoints_boundary = (12**2, None)
```

```python
# cohesive energy versus k-points for c-Beryllene bulk

# kpoints_boundary = (10**3, None)

summarize_cohesive_energy("1.0_convergence/c0-Beryllene_bulk","1.0_convergence/single_Beryllium")
t_kpoints = ["c-Beryllene bulk", "1.0_convergence/c0-Beryllene_bulk/cohesive_energy.dat", kpoints_boundary, "blue"]
kpoints_list = [t_kpoints]

plot_cohesive_energy_parameters("Cohesive energy versus k-points for c-Beryllene bulk", "kpoints", kpoints_list)
plt.savefig("figures/2.cohesive_c0-Beryllene.pdf")
```

```python
# cohesive energy versus k-points for c-Beryllene monolayer

summarize_cohesive_energy("1.0_convergence/c1-Beryllene_monolayer","1.0_convergence/single_Beryllium")
c_kpoints = ["c-Beryllene", "1.0_convergence/c1-Beryllene_monolayer/cohesive_energy.dat", kpoints_boundary, "orange"]
kpoints_list = [c_kpoints]

plot_cohesive_energy_parameters("Cohesive energy versus k-points for c-Beryllene monolayer", "kpoints", kpoints_list)
plt.savefig("figures/2.cohesive_c1-Beryllene.pdf")
```

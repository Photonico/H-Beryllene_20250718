# Cohesive energy testing for cubic family

Cohesive energy versus parameters


```python
# prework

from vmatplot.convergence import summarize_cohesive_energy, plot_cohesive_energy_parameters
import matplotlib.pyplot as plt

encut_boundary = (300, None)
kpoints_boundary = (12**2, None)
```

```python
# cohesive energy versus k-points for bulk bcc beryllium

# kpoints_boundary = (10**3, None)

summarize_cohesive_energy("1.0_convergence/c0-Beryllium","1.0_convergence/single_Beryllium")
t_kpoints = ["bulk bcc beryllium", "1.0_convergence/c0-Beryllium/cohesive_energy.dat", kpoints_boundary, "blue"]
kpoints_list = [t_kpoints]

plot_cohesive_energy_parameters("Cohesive energy versus k-points for bulk bcc beryllium", "kpoints", kpoints_list)
plt.savefig("figures/1_convergence/2.5_coh_c0.pdf")
plt.savefig("figures_for_publication/S3.1c_conv_bcc.pdf")
```

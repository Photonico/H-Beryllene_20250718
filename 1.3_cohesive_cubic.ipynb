# Cohesive energy testing for cubic family

Cohesive energy versus parameters


```python
# prework

from vmatplot.convergence import summarize_cohesive_energy, plot_cohesive_energy_parameters
import matplotlib.pyplot as plt
from vmatplot.output_settings import figure_version, save_figure

figure_version("article")    # "thesis": thesis typography, figures saved as figures/<name>_thesis.pdf

encut_boundary = (300, None)
kpoints_boundary = (12**2, None)
```

```python
# cohesive energy versus k-points for bulk bcc beryllium

# kpoints_boundary = (10**3, None)

summarize_cohesive_energy("1.0_convergence/c0-Beryllium","1.0_convergence/single_Beryllium")
t_kpoints = ["bulk bcc beryllium", "1.0_convergence/c0-Beryllium/cohesive_energy.dat", kpoints_boundary, "blue"]
kpoints_list = [t_kpoints]

plot_cohesive_energy_parameters("Cohesive energy versus k-points for bulk bcc beryllium", "kpoints", kpoints_list)
save_figure("convergence_cohesive_kpoints_bcc-beryllium")
```

```python
# cohesive energy versus k-points for c-beryllene monolayer

summarize_cohesive_energy("1.0_convergence/c1-Beryllene_monolayer","1.0_convergence/single_Beryllium")
c_kpoints = ["c-beryllene", "1.0_convergence/c1-Beryllene_monolayer/cohesive_energy.dat", kpoints_boundary, "blue"]
kpoints_list = [c_kpoints]

plot_cohesive_energy_parameters("Cohesive energy versus k-points for c-beryllene monolayer", "kpoints", kpoints_list)
save_figure("convergence_cohesive_kpoints_cubic-beryllene-monolayer")
```

```python
# cohesive energy versus k-points for c-beryllene bilayer

summarize_cohesive_energy("1.0_convergence/c2-Beryllene_bilayer","1.0_convergence/single_Beryllium")
t_kpoints = ["c-beryllene bilayer", "1.0_convergence/c2-Beryllene_bilayer/cohesive_energy.dat", kpoints_boundary, "blue"]
kpoints_list = [t_kpoints]

plot_cohesive_energy_parameters("Cohesive energy versus k-points for c-beryllene bilayer", "kpoints", kpoints_list)
save_figure("convergence_cohesive_kpoints_cubic-beryllene-bilayer")
```

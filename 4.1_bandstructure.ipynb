# Band structure

```python
# prework

from vmatplot.bandstructure import plot_bandstructure
import matplotlib.pyplot as plt
```

```python
# α-Beryllium bulk

bs_list = [["monocolor", "α-Beryllium", "4.0_bandstructure/a-Beryllium", "blue"]]

plot_bandstructure("Bandstructure for α-Beryllium bulk", bs_list, (-8,8))
plt.savefig("figures/4.10_BS_a-Beryllium.pdf")
```

```python
# α-Beryllene

bs_list = [["monocolor", "α-Beryllene", "4.0_bandstructure/a-Beryllene", "blue"]]

plot_bandstructure("Bandstructure for α-Beryllene", bs_list, (-8,8))
plt.savefig("figures/4.11_BS_a-Beryllene.pdf")
```

```python
# β-Beryllene

bs_list = [["monocolor", "β-Beryllene", "4.0_bandstructure/b-Beryllene", "blue"]]

plot_bandstructure("Bandstructure for β-Beryllene", bs_list, (-8,8))
plt.savefig("figures/4.12_BS_b-Beryllene.pdf")
```

```python
# c-Beryllium

bs_list = [["monocolor", "c-Beryllium bulk", "4.0_bandstructure/c0-Beryllium", "violet"]]

plot_bandstructure("Bandstructure for cubic Beryllium bulk", bs_list, (-8,8))
plt.savefig("figures/4.20_BS_c-Beryllium.pdf")
```

```python
# c-Beryllene trilayer

bs_list = [["monocolor", "c-Beryllene trilayer", "4.0_bandstructure/c3-Beryllene_trilayer", "violet"]]

plot_bandstructure("Bandstructure for cubic Beryllene trilayer", bs_list, (-8,8))
plt.savefig("figures/4.21_BS_c-Beryllene_trilayer.pdf")
```

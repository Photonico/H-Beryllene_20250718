# Band structure

```python
# prework

from vmatplot.bandstructure import plot_bandstructure
import matplotlib.pyplot as plt
```

```python
# α-Beryllium bulk

bs_list = [["monocolor", "α-Beryllium", "4.0_bandstructure/a0-Beryllium", "purple"]]

plot_bandstructure("Bandstructure for α-Beryllium bulk", bs_list, (-8,8))
plt.savefig("figures/3.41_bs_a-Beryllium.pdf")
```

```python
# c-Beryllene bulk

bs_list = [["monocolor", "c-Beryllene bulk", "4.0_bandstructure/c0-Beryllene_bulk_alt", "cyan"]]

plot_bandstructure("Bandstructure for cubic Beryllene bulk", bs_list, (-8,8))
plt.savefig("figures/3.44_bs_c_bulk.pdf")
```

```python
# c-Beryllene trilayer

bs_list = [["monocolor", "c-Beryllene trilayer", "4.0_bandstructure/c3-Beryllene_trilayer", "orange"]]

plot_bandstructure("Bandstructure for cubic Beryllene trilayer", bs_list, (-8,8))
plt.savefig("figures/3.47_bs_c_trilayer.pdf")
```

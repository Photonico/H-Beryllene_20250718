# Band structure

```python
# prework

from vmatplot.bandstructure import plot_bandstructure
import matplotlib.pyplot as plt
```

```python
# Bandstructure a- g-

bs_list = [["monocolor", "α-Beryllene monolayer", "4.0_bandstructure/a1-Beryllene_monolayer", "purple"],
           ["monocolor", "g-Beryllene monolayer", "4.0_bandstructure/g-Beryllene_monolayer", "blue"]
           ]

plot_bandstructure("Bandstructure for hexagonal Beryllene systems", bs_list, (-4,4), True)
plt.savefig("figures/3.40_phonon_ag_layers.pdf")
```

```python
# Bandstructure c-

bs_list = [# ["monocolor", "c-Beryllene bulk", "4.0_bandstructure/c0-Beryllene_bulk", "cyan"],
           ["monocolor", "c-Beryllene monolayer", "4.0_bandstructure/c1-Beryllene_monolayer", "lime"],
           ["monocolor", "c-Beryllene bilayer", "4.0_bandstructure/c2-Beryllene_bilayer", "yellow"],
           ["monocolor", "c-Beryllene trilayer", "4.0_bandstructure/c3-Beryllene_trilayer", "orange"],
           ]

plot_bandstructure("Bandstructure for cubic Beryllene systems", bs_list, (-8,8), True)
plt.savefig("figures/3.41_phonon_c_layers.pdf")
```

# Cohesive energy testing for hexagonal family

Cohesive energy versus parameters


```python
# prework

from vmatplot.convergence import summarize_cohesive_energy, plot_cohesive_energy_parameters
import matplotlib.pyplot as plt

encut_boundary = (300, None)
kpoints_boundary = (12**2, None)
```

```python
# cohesive energy versus energy cutoff

summarize_cohesive_energy("1.0_convergence/g-Beryllene_energycutoff","1.0_convergence/single_Beryllium")
g_encut = ["g-Beryllene monolayer", "1.0_convergence/g1-Beryllene_energycutoff/cohesive_energy.dat", encut_boundary, "yellow"]
encut_list = [g_encut]

plot_cohesive_energy_parameters("Cohesive energy versus energy cutoff for g-Beryllene monolayer", "encut", encut_list )
plt.savefig("figures/2.cohesive_a1-Beryllene_encut.pdf")
```

```python
# cohesive energy versus k-points for a-Beryllium

summarize_cohesive_energy("1.0_convergence/a0-Beryllium","1.0_convergence/single_Beryllium")
a_kpoints = ["α-Beryllium", "1.0_convergence/a0-Beryllium/cohesive_energy.dat", kpoints_boundary, "blue"]
kpoints_list = [a_kpoints]

plot_cohesive_energy_parameters("Cohesive energy versus k-points for α-Beryllium", "kpoints", kpoints_list)
plt.savefig("figures/2.cohesive_a0-Beryllium.pdf")
```

```python
# cohesive energy versus k-points for a-Beryllene

summarize_cohesive_energy("1.0_convergence/a1-Beryllene_monolayer","1.0_convergence/single_Beryllium")
a_kpoints = ["α-Beryllene monolayer", "1.0_convergence/a1-Beryllene_monolayer/cohesive_energy.dat", kpoints_boundary, "orange"]
kpoints_list = [a_kpoints]

plot_cohesive_energy_parameters("Cohesive energy versus k-points for α-Beryllene monolayer", "kpoints", kpoints_list)
plt.savefig("figures/2.cohesive_a1-Beryllene.pdf")
```

```python
# cohesive energy versus k-points for g-Beryllene

summarize_cohesive_energy("1.0_convergence/g-Beryllene_kpoints","1.0_convergence/single_Beryllium")
g_kpoints = ["g-Beryllene monolayer", "1.0_convergence/g1-Beryllene_kpoints/cohesive_energy.dat", kpoints_boundary, "yellow"]
kpoints_list = [g_kpoints]

plot_cohesive_energy_parameters("Cohesive energy versus k-points for g-Beryllene monolayer", "kpoints", kpoints_list)
plt.savefig("figures/2.cohesive_g1-Beryllene.pdf")
```

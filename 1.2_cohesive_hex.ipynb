# Cohesive energy testing for hexagonal family

Cohesive energy versus parameters


```python
# prework

from vmatplot.convergence import summarize_cohesive_energy, plot_cohesive_energy_parameters
import matplotlib.pyplot as plt

encut_boundary = (300, None)
kpoints_boundary = (12**2, None)
```

```python
# cohesive energy versus energy cutoff for g-Beryllene

summarize_cohesive_energy("1.0_convergence/g-Beryllene_energycutoff","1.0_convergence/single_Beryllium")
g_encut = ["g-Beryllene monolayer", "1.0_convergence/g-Beryllene_energycutoff/cohesive_energy.dat", encut_boundary, "blue"]
encut_list = [g_encut]

plot_cohesive_energy_parameters("Cohesive energy versus energy cutoff for graphic Beryllene monolayer", "encut", encut_list )
plt.savefig("figures/2.1_coh_g_encut.pdf")
```

```python
# cohesive energy versus k-points for g-Beryllene

summarize_cohesive_energy("1.0_convergence/g-Beryllene_kpoints","1.0_convergence/single_Beryllium")
g_kpoints = ["g-Beryllene monolayer", "1.0_convergence/g-Beryllene_kpoints/cohesive_energy.dat", kpoints_boundary, "blue"]
kpoints_list = [g_kpoints]

plot_cohesive_energy_parameters("Cohesive energy versus k-points for graphic Beryllene monolayer", "kpoints", kpoints_list)
plt.savefig("figures/2.2_coh_g_kpoints.pdf")
```

```python
# cohesive energy versus k-points for a-Beryllium

summarize_cohesive_energy("1.0_convergence/a-Beryllium","1.0_convergence/single_Beryllium")
a_kpoints = ["α-Beryllium", "1.0_convergence/a-Beryllium/cohesive_energy.dat", kpoints_boundary, "blue"]
kpoints_list = [a_kpoints]

plot_cohesive_energy_parameters("Cohesive energy versus k-points for α-Beryllium", "kpoints", kpoints_list)
plt.savefig("figures/2.3_coh_a0.pdf")
```

```python
# cohesive energy versus k-points for a-Beryllene

summarize_cohesive_energy("1.0_convergence/a-Beryllene","1.0_convergence/single_Beryllium")
a_kpoints = ["α-Beryllene monolayer", "1.0_convergence/a-Beryllene/cohesive_energy.dat", kpoints_boundary, "blue"]
kpoints_list = [a_kpoints]

plot_cohesive_energy_parameters("Cohesive energy versus k-points for α-Beryllene monolayer", "kpoints", kpoints_list)
plt.savefig("figures/2.4_coh_a1.pdf")
```

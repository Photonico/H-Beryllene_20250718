# Phonon dispersion for hexagonal family


```python
# prework

from vmatplot.phonon import plot_phonons
import matplotlib.pyplot as plt

y_range = [-20, 30]
```

```python
# α-Beryllene

matters = [["α-Beryllene", "3.0_phonon_dispersion/a-Beryllene", "violet", "solid", 1.5, 1.0, 0]]

plot_phonons("Phonon dispersion for α-Beryllene", matters, y_range, None)
plt.savefig("figures/3.12_phonon_a-Beryllene.pdf")
```

```python
# β-Beryllene

matters = [["α-Beryllene", "3.0_phonon_dispersion/b-Beryllene", "blue", "solid", 1.5, 1.0, 0]]

plot_phonons("Phonon dispersion for β-Beryllene", matters, y_range, None)
plt.savefig("figures/3.13_phonon_b-Beryllene.pdf")
```

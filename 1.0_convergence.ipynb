# Convergence testing

Energy versus parameters


```python
# prework

from vmatplot.convergence import summarize_parameters, plot_energy_parameters
import matplotlib.pyplot as plt

kpoints_boundary = (16, None)
encut_boundary = (250, None)
```

```python
# energy versus energy cutoff

summarize_parameters("1.0_convergence/G-Beryllene_energycutoff")
G_encut = ["for g-Beryllene", "1.0_convergence/G-Beryllene_energycutoff/energy_parameters.dat", encut_boundary, "blue"]
encut_list = [G_encut]

plot_energy_parameters("Energy versus energy cutoff", "encut", encut_list)
plt.savefig("figures/1.0a_encut.pdf")
```

```python
# energy versus kpoints

summarize_parameters("1.0_convergence/G-Beryllene_kpoints")
G_kpoints = ["for g-Beryllene", "1.0_convergence/G-Beryllene_kpoints/energy_parameters.dat", kpoints_boundary, "violet"]
kpoints_list = [G_kpoints]

plot_energy_parameters("Energy versus energy cutoff", "kpoints", kpoints_list)
plt.savefig("figures/1.0b_kpoints.pdf")
```

# Convergence testing

Energy versus parameters


```python
# prework

from vmatplot.convergence import summarize_parameters, plot_energy_parameters
import matplotlib.pyplot as plt

encut_boundary = (300, None)
kpoints_boundary = (12**2, None)
```

```python
# energy versus energy cutoff

summarize_parameters("1.0_convergence/g-Beryllene_energycutoff")
g_encut = ["g-Beryllene", "1.0_convergence/g-Beryllene_energycutoff/energy_parameters.dat", encut_boundary, "blue"]
encut_list = [g_encut]

plot_energy_parameters("Energy versus energy cutoff for g-Beryllene", "encut", encut_list)
plt.savefig("figures/1.00_encut.pdf")
```

```python
# energy versus k-points for a-Beryllene

summarize_parameters("1.0_convergence/a-Beryllene_kpoints")
a_kpoints = ["a-Beryllene", "1.0_convergence/a-Beryllene_kpoints/energy_parameters.dat", kpoints_boundary, "violet"]
kpoints_list = [a_kpoints]

plot_energy_parameters("Energy versus k-points for a-Beryllene", "kpoints", kpoints_list)
plt.savefig("figures/1.01_kpoints_a.pdf")
```

```python
# energy versus k-points for g-Beryllene

summarize_parameters("1.0_convergence/g-Beryllene_kpoints")
g_kpoints = ["g-Beryllene", "1.0_convergence/g-Beryllene_kpoints/energy_parameters.dat", kpoints_boundary, "blue"]
kpoints_list = [g_kpoints]

plot_energy_parameters("Energy versus k-points for g-Beryllene", "kpoints", kpoints_list)
plt.savefig("figures/1.02_kpoints_g.pdf")
```

```python
# energy versus k-points for c-Beryllene

summarize_parameters("1.0_convergence/c-Beryllene_kpoints")
c_kpoints = ["c-Beryllene", "1.0_convergence/c-Beryllene_kpoints/energy_parameters.dat", kpoints_boundary, "lime"]
kpoints_list = [c_kpoints]

plot_energy_parameters("Energy versus k-points for c-Beryllene", "kpoints", kpoints_list)
plt.savefig("figures/1.03_kpoints_c.pdf")
```

```python
# energy versus k-points for c-Beryllene trilayer

summarize_parameters("1.0_convergence/c-Beryllene_trilayer_kpoints")
t_kpoints = ["c-Beryllene trilayer", "1.0_convergence/c-Beryllene_trilayer_kpoints/energy_parameters.dat", kpoints_boundary, "yellow"]
kpoints_list = [t_kpoints]

plot_energy_parameters("Energy versus k-points for c-Beryllene trilayer", "kpoints", kpoints_list)
plt.savefig("figures/1.04_kpoints_t.pdf")
```

# Convergence testing for hexagonal family

Energy versus parameters


```python
# prework

from vmatplot.convergence import summarize_parameters, plot_energy_parameters
import matplotlib.pyplot as plt
from vmatplot.output_settings import figure_version, save_figure

figure_version("article")    # "thesis": thesis typography, figures saved as figures/<name>_thesis.pdf

encut_boundary = (300, None)
kpoints_boundary = (12**2, None)
```

```python
# energy versus energy cutoff for g-beryllene

summarize_parameters("1.0_convergence/g-Beryllene_energycutoff")
g_encut = ["g-beryllene monolayer", "1.0_convergence/g-Beryllene_energycutoff/energy_parameters.dat", encut_boundary, "blue"]
encut_list = [g_encut]

plot_energy_parameters("Energy versus energy cutoff for graphic Beryllene monolayer", "encut", encut_list)
save_figure("convergence_energy_encut_g-beryllene")
```

```python
# energy versus k-points for g-beryllene

summarize_parameters("1.0_convergence/g-Beryllene_kpoints")
g_kpoints = ["g-beryllene monolayer", "1.0_convergence/g-Beryllene_kpoints/energy_parameters.dat", kpoints_boundary, "blue"]
kpoints_list = [g_kpoints]

plot_energy_parameters("Energy versus k-points for graphic Beryllene monolayer", "kpoints", kpoints_list)
save_figure("convergence_energy_kpoints_g-beryllene")
```

```python
# cohesive energy versus k-points for a-Beryllium

summarize_parameters("1.0_convergence/a-Beryllium")
a_kpoints = ["bulk hcp beryllium", "1.0_convergence/a-Beryllium/energy_parameters.dat", kpoints_boundary, "blue"]
kpoints_list = [a_kpoints]

plot_energy_parameters("Energy versus k-points for bulk hcp beryllium", "kpoints", kpoints_list)
save_figure("convergence_energy_kpoints_hcp-beryllium")
```

```python
# cohesive energy versus k-points for a-beryllene

summarize_parameters("1.0_convergence/a-Beryllene")
a_kpoints = ["α-beryllene", "1.0_convergence/a-Beryllene/energy_parameters.dat", kpoints_boundary, "blue"]
kpoints_list = [a_kpoints]

plot_energy_parameters("Energy versus k-points for α-beryllene", "kpoints", kpoints_list)
save_figure("convergence_energy_kpoints_alpha-beryllene")
```

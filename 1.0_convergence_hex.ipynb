# Convergence testing for hexagonal family

Energy versus parameters


```python
# prework

from vmatplot.convergence import summarize_parameters, plot_energy_parameters
import matplotlib.pyplot as plt

encut_boundary = (300, None)
kpoints_boundary = (12**2, None)
```

```python
# energy versus energy cutoff for g-Beryllene

summarize_parameters("1.0_convergence/g-Beryllene_energycutoff")
g_encut = ["g-Beryllene monolayer", "1.0_convergence/g1-Beryllene_energycutoff/energy_parameters.dat", encut_boundary, "yellow"]
encut_list = [g_encut]

plot_energy_parameters("Energy versus energy cutoff for g-Beryllene monolayer", "encut", encut_list)
plt.savefig("figures/1.conv_a1-Beryllene_encut.pdf")
```

```python
# energy versus k-points for a-Beryllium

summarize_parameters("1.0_convergence/a0-Beryllium")
a_kpoints = ["α-Beryllium", "1.0_convergence/a0-Beryllium/energy_parameters.dat", kpoints_boundary, "blue"]
kpoints_list = [a_kpoints]

plot_energy_parameters("Energy versus k-points for α-Beryllium", "kpoints", kpoints_list)
plt.savefig("figures/1.conv_a0-Beryllium.pdf")
```

```python
# energy versus k-points for a-Beryllene

summarize_parameters("1.0_convergence/a1-Beryllene_monolayer")
a_kpoints = ["α-Beryllene monolayer", "1.0_convergence/a1-Beryllene_monolayer/energy_parameters.dat", kpoints_boundary, "orange"]
kpoints_list = [a_kpoints]

plot_energy_parameters("Energy versus k-points for α-Beryllene monolayer", "kpoints", kpoints_list)
plt.savefig("figures/1.conv_a1-Beryllene.pdf")
```

```python
# energy versus k-points for g-Beryllene

summarize_parameters("1.0_convergence/g-Beryllene_kpoints")
g_kpoints = ["g-Beryllene monolayer", "1.0_convergence/g1-Beryllene_kpoints/energy_parameters.dat", kpoints_boundary, "yellow"]
kpoints_list = [g_kpoints]

plot_energy_parameters("Energy versus k-points for g-Beryllene monolayer", "kpoints", kpoints_list)
plt.savefig("figures/1.conv_g1-Beryllene.pdf")
```

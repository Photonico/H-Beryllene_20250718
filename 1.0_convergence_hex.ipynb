# Convergence testing

Energy versus parameters


```python
# prework

from vmatplot.convergence import summarize_parameters, plot_energy_parameters
import matplotlib.pyplot as plt

encut_boundary = (300, None)
kpoints_boundary = (12**2, None)
```

```python
# energy versus energy cutoff for g-Beryllene

summarize_parameters("1.0_convergence/g-Beryllene_energycutoff")
g_encut = ["g-Beryllene monolayer", "1.0_convergence/g-Beryllene_energycutoff/energy_parameters.dat", encut_boundary, "blue"]
encut_list = [g_encut]

plot_energy_parameters("Energy versus energy cutoff for g-Beryllene monolayer", "encut", encut_list)
plt.savefig("figures/1.00_encut.pdf")
```

```python
# energy versus k-points for a-Beryllium

summarize_parameters("1.0_convergence/a0-Beryllium")
a_kpoints = ["α-Beryllium", "1.0_convergence/a0-Beryllium/energy_parameters.dat", kpoints_boundary, "purple"]
kpoints_list = [a_kpoints]

plot_energy_parameters("Energy versus k-points for α-Beryllium", "kpoints", kpoints_list)
plt.savefig("figures/1.01_conv_a-Beryllium.pdf")
```

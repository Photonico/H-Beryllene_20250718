# Density of State


```python
# prework

from vmatplot.dos import plot_dos
import matplotlib.pyplot as plt
```

```python
# overview DoS

systems = [["α-Beryllium bulk",         "5.0_PDoS/a-Beryllium",     "unpolarized",          "violet"],
           ["α-beryllene",              "5.0_PDoS/a-beryllene",     "unpolarized",          "blue"],
           ["β-beryllene",              "5.0_PDoS/b-beryllene",     "unpolarized",          "lime"],
           ["cubic Beryllium bulk",     "5.0_PDoS/c0-Beryllium",    "unpolarized",          "purple"],
           ["cubic beryllene trilayer", "5.0_PDoS/c3-beryllene",    "unpolarized",          "orange"]]

plot_dos("DoS", systems, 8, [0, 4], "Total")
plt.savefig("figures/5_dos/5.10_DoS_overview.pdf")
```

```python
# hexagonal family

systems = [["α-Beryllium bulk",         "5.0_PDoS/a-beryllene",     "unpolarized",  "violet"],
           ["α-beryllene",              "5.0_PDoS/a-Beryllium",     "unpolarized",  "blue"],
           ["β-beryllene",              "5.0_PDoS/b-beryllene",     "unpolarized",  "lime"],]

plot_dos("DoS for hexagonal family", systems, 8, [0, 4], "Total")
plt.savefig("figures/5_dos/5.11_DoS_hex.pdf")
```

```python
# cubic family

systems = [["bulk bcc beryllium",     "5.0_PDoS/c0-Beryllium",    "unpolarized",  "purple"],
           ["cubic beryllene trilayer", "5.0_PDoS/c3-beryllene",    "unpolarized",  "orange"]]

plot_dos("DoS for cubic family", systems, 8, [0, 4], "Total")
plt.savefig("figures/5_dos/5.12_DoS_cubic.pdf")
plt.savefig("figures_for_publication/fig3.11_DoS_cubic.pdf")
```

# Projected Density of State


```python
# prework

from vmatplot.pdos import plot_single_pdos
import matplotlib.pyplot as plt
```

```python
# α-Beryllene hh

be_index = [1]
h_index = [2,3]
current_dict = "5.0_PDoS/a-Beryllene_hh"

systems = [
           ["Total DOS", current_dict, [], "total_dos", "silver", "solid", 2.0, 1.0],
           ["Total PDoS", current_dict, [], "total", "grey", "solid", 2.0, 1.0],
           ["$p$ for Be", current_dict, be_index, "p", "blue", "solid", 2.0, 1.0],
           ["$s$ for Be", current_dict, be_index, "s", "blue", "dashed", 2.0, 1.0],
           ["$p$ for H", current_dict, h_index, "p", "orange", "solid", 2.0, 1.0],
           ["$s$ for H", current_dict, h_index, "s", "orange", "dashed", 2.0, 1.0],
           ]

plot_single_pdos("PDoS for α-Beryllene (double side Hydrogen adhesion)", systems, 4, 3)
plt.savefig("figures/5_pdos/5.31_a_hh.pdf")
```

```python
# β-Beryllene h

be_upper_index = [1]
be_lower_index = [2]
h_index = [3]
current_dict = "5.0_PDoS/b-Beryllene_h"

systems = [
           ["Total DOS", current_dict, [], "total_dos", "silver", "solid", 2.0, 1.0],
           ["Total PDoS", current_dict, [], "total", "grey", "solid", 2.0, 1.0],
           ["$p$ for upper Be", current_dict, be_upper_index, "p", "blue", "solid", 2.0, 1.0],
           ["$s$ for upper Be", current_dict, be_upper_index, "s", "blue", "dashed", 2.0, 1.0],
           ["$p$ for lower Be", current_dict, be_lower_index, "p", "violet", "solid", 2.0, 1.0],
           ["$s$ for lower Be", current_dict, be_lower_index, "s", "violet", "dashed", 2.0, 1.0],
           ["$p$ for H", current_dict, h_index, "p", "orange", "solid", 2.0, 1.0],
           ["$s$ for H", current_dict, h_index, "s", "orange", "dashed", 2.0, 1.0],
           ]

plot_single_pdos("PDoS for β-Beryllene (single side Hydrogen adhesion)", systems, 4, 1.5)
plt.savefig("figures/5_pdos/5.32_b_h.pdf")
```

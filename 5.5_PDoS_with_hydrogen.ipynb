# Projected Density of State


```python
# prework

from vmatplot.pdos import plot_single_pdos
import matplotlib.pyplot as plt
from vmatplot.output_settings import figure_version, save_figure

figure_version("article")    # "thesis": thesis typography, figures saved as figures/<name>_thesis.pdf
```

```python
# α-beryllene hh

be_index = [1]
h_index = [2,3]
current_dict = "5.0_PDoS/a-Beryllene_hh"

systems = [
           ["Total DOS", current_dict, [], "total_dos", "silver", "solid", 1.5, 1.0],
           ["Total PDoS", current_dict, [], "total", "grey", "solid", 1.5, 1.0],
           ["$p$ for Be", current_dict, be_index, "p", "blue", "solid", 1.5, 1.0],
           ["$s$ for Be", current_dict, be_index, "s", "blue", "dashed", 1.5, 1.0],
           ["$p$ for H", current_dict, h_index, "p", "orange", "solid", 1.5, 1.0],
           ["$s$ for H", current_dict, h_index, "s", "orange", "dashed", 1.5, 1.0],
           ]

plot_single_pdos("PDoS for double-sided H-terminated α-beryllene", systems, 4, 3)
save_figure("pdos_2H-alpha-beryllene")
```

```python
# β-beryllene h

be_upper_index = [1]
be_lower_index = [2]
h_index = [3]
current_dict = "5.0_PDoS/b-Beryllene_b"

systems = [
           ["Total DOS", current_dict, [], "total_dos", "silver", "solid", 1.5, 1.0],
           ["Total PDoS", current_dict, [], "total", "grey", "solid", 1.5, 1.0],
           ["$p$ for upper Be", current_dict, be_upper_index, "p", "blue", "solid", 1.5, 1.0],
           ["$s$ for upper Be", current_dict, be_upper_index, "s", "blue", "dashed", 1.5, 1.0],
           ["$p$ for lower Be", current_dict, be_lower_index, "p", "violet", "solid", 1.5, 1.0],
           ["$s$ for lower Be", current_dict, be_lower_index, "s", "violet", "dashed", 1.5, 1.0],
           ["$p$ for H", current_dict, h_index, "p", "orange", "solid", 1.5, 1.0],
           ["$s$ for H", current_dict, h_index, "s", "orange", "dashed", 1.5, 1.0],
           ]

plot_single_pdos("PDoS for H-terminated β-beryllene", systems, 4, 1.5)
save_figure("pdos_1H-beta-beryllene")
```

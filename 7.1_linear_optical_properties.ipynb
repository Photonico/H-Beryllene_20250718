# Linear Optical Properties


```python
# prework

from vmatplot.linear_optical_properties import plot_linear_optical_property
import matplotlib.pyplot as plt

components = [{"xx":"xx-component"}, {"yy":"yy-component"}, {"zz":"zz-component"}]
photon_boundary = (0, 12)
layout = "hor"

# System factors (d_c, (z_max-z_min)*d_c+2*r_be) r_be = 1.98
factor_bulk = (1, 1)                                        # for α-Beryllium and cubic Beryllium bulk
factor_a = (40, 2*1.98)                                     # for α-Beryllene monolayer
factor_b = (40, 0.0473771641975619+2*1.98)                  # for β-Beryllene monolayer
factor_c = (40, 0.0751622445053417+2*1.98)                  # for cubic Beryllene trilayer

# systems list
systems = [
           ["bulk hcp Beryllium",       "6.0_dielectric_selection/a-Beryllium",  "violet",   None,None,None, factor_bulk],
           ["α-Beryllene",              "6.0_dielectric_selection/a-Beryllene",  "blue",     None,None,None, factor_a],
           ["β-Beryllene",              "6.0_dielectric_selection/b-Beryllene",  "lime",     None,None,None, factor_b],
           ["bulk bcc Beryllium",       "6.0_dielectric_selection/c0-Beryllium", "purple",   None,None,None, factor_bulk],
           ["cubic Beryllene trilayer", "6.0_dielectric_selection/c3-Beryllene", "orange",   None,None,None, factor_c]
           ]
```

```python
# Absorption coefficient

lim = 0.05
value_boundary = (-0.05*lim, 1.05*lim)
plot_linear_optical_property("Absorption coefficient", systems, "absorption", components, layout, True, "eV", photon_boundary, value_boundary, True)
plt.savefig("figures/7_optical/7.11_absorption.pdf")
plt.savefig("figures_for_publication/fig3.17a_abs.pdf")
```

```python
# Refractive index

lim = 7.5
value_boundary = (-0.05*lim, 1.05*lim)
plot_linear_optical_property("Refractive index", systems, "refractive", components, layout, True, "eV", photon_boundary, value_boundary, True)
plt.savefig("figures/7_optical/7.12_refractive.pdf")
plt.savefig("figures_for_publication/S3.18a_refractive.pdf")
```

```python
# Extinction coefficient

lim = 5
value_boundary = (-0.05*lim, 1.05*lim)
plot_linear_optical_property("Extinction coefficient", systems, "extinction", components, layout, True, "eV", photon_boundary, value_boundary, True)
plt.savefig("figures/7_optical/7.13_extinction.pdf")
plt.savefig("figures_for_publication/S3.18c_extinction.pdf")
```

```python
# Reflectivity

lim = 1
value_boundary = (-0.05*lim, 1.05*lim)
plot_linear_optical_property("Reflectivity", systems, "reflectivity", components, layout, True, "eV", photon_boundary, value_boundary, True)
plt.savefig("figures/7_optical/7.14_reflectivity.pdf")
plt.savefig("figures_for_publication/S3.18b_reflectivity.pdf")
```

```python
# Energy-loss spectrum

lim = 2
value_boundary = (-0.05*lim, 1.05*lim)
plot_linear_optical_property("Energy-loss spectrum", systems, "energy-loss", components, layout, True, "eV", photon_boundary, value_boundary, True)
plt.savefig("figures/7_optical/7.15_energy-loss.pdf")
plt.savefig("figures_for_publication/fig3.17b_energy-loss.pdf")
```

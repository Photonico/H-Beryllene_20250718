# Linear Optical Properties

Optical normalization uses the relaxed structure of each calculation and the original surface-radius convention (Be: 1.98 Å; H: 0.70 Å). Hydrogenated slabs use the species at the outermost atomic centres; this convention is not a union of atomic spheres. Both xx and zz retain the same effective-volume normalization. Absorption is reported in nm$^{-1}$ with $\omega=2\pi E/h$. These are effective-thickness IPA interband spectra; no Drude term is added. Scalar reflectivity uses the semi-infinite-medium expression, not a finite-sheet measurement model.

```python
# prework

from vmatplot.linear_optical_properties import plot_linear_optical_property
from vmatplot.structure import slab_optical_factor
from vmatplot.plot_validation import ensure_curve_visibility
import matplotlib.pyplot as plt
from vmatplot.output_settings import figure_version, save_figure

figure_version("article")    # "thesis": thesis typography, figures saved as figures/<name>_thesis.pdf

components = [{"xx":"xx-component"}, {"yy":"yy-component"}, {"zz":"zz-component"}]
photon_boundary = (0, 12)
layout = "hor"

# Optical factors from each relaxed structure; radii: Be = 1.98 A, H = 0.70 A.
# The cell height is the full surface-normal period, not the vacuum gap.
factor_bulk = (1, 1)                                        # for α-Beryllium and cubic Beryllium bulk
factor_a = slab_optical_factor("6.0_dielectric_selection/a-Beryllene")
factor_b = slab_optical_factor("6.0_dielectric_selection/b-Beryllene_p3m1")
factor_c = slab_optical_factor("6.0_dielectric_selection/c3-Beryllene")

# systems list
systems = [
           ["bulk hcp beryllium",       "6.0_dielectric_selection/a-Beryllium",  "violet",   None,None,None, factor_bulk],
           ["α-beryllene",              "6.0_dielectric_selection/a-Beryllene",  "blue",     None,None,None, factor_a],
           ["β-beryllene",              "6.0_dielectric_selection/b-Beryllene_p3m1",  "lime",     None,None,None, factor_b],
           ["bulk bcc beryllium",       "6.0_dielectric_selection/c0-Beryllium", "purple",   None,None,None, factor_bulk],
           ["trilayer cubic beryllene", "6.0_dielectric_selection/c3-Beryllene", "orange",   None,None,None, factor_c]
           ]
```

```python
# Absorption coefficient

value_boundary = (None, None)  # Fit each panel to all curves in the energy window.
plot_linear_optical_property("Absorption coefficient", systems, "absorption", components, layout, True, "eV", photon_boundary, value_boundary, True)
ensure_curve_visibility(plt.gcf())
save_figure("optics_pristine_absorption")
```

```python
# Refractive index

lim = 7.5
value_boundary = (-0.05*lim, 1.05*lim)
plot_linear_optical_property("Refractive index", systems, "refractive", components, layout, True, "eV", photon_boundary, value_boundary, True)
ensure_curve_visibility(plt.gcf())
save_figure("optics_pristine_refractive")
```

```python
# Extinction coefficient

lim = 5
value_boundary = (-0.05*lim, 1.05*lim)
plot_linear_optical_property("Extinction coefficient", systems, "extinction", components, layout, True, "eV", photon_boundary, value_boundary, True)
ensure_curve_visibility(plt.gcf())
save_figure("optics_pristine_extinction")
```

```python
# Reflectivity

lim = 1
value_boundary = (-0.05*lim, 1.05*lim)
plot_linear_optical_property("Reflectivity", systems, "reflectivity", components, layout, True, "eV", photon_boundary, value_boundary, True)
ensure_curve_visibility(plt.gcf())
save_figure("optics_pristine_reflectivity")
```

```python
# Energy-loss spectrum

value_boundary = (None, None)  # Fit each panel to all curves in the energy window.
plot_linear_optical_property("Energy-loss spectrum", systems, "energy-loss", components, layout, True, "eV", photon_boundary, value_boundary, True)
ensure_curve_visibility(plt.gcf())
save_figure("optics_pristine_energy-loss")
```

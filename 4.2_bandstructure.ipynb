# Band structure

```python
# prework

from vmatplot.bandstructure import plot_bandstructure
import matplotlib.pyplot as plt
from vmatplot.output_settings import figure_version, save_figure

figure_version("article")    # "thesis": thesis typography, figures saved as figures/<name>_thesis.pdf

# Zero of energy: the Fermi energy of the self-consistent run (last entry of each bs_list), not that of the
# non-self-consistent band run, whose E-fermi is obtained from the path k-points only
```

```python
# α-Beryllium bulk

bs_list = [["monocolor", "bulk hcp beryllium", "4.0_bandstructure/a-Beryllium", "blue", "solid", 1.5, 1.0, 0, "2.4_self_consistent/a-Beryllium"]]

plot_bandstructure("bulk hcp beryllium", bs_list, (-8,8))
save_figure("bands_hcp-beryllium")
```

# Band structure

```python
# prework

from vmatplot.bandstructure import plot_bandstructure
import matplotlib.pyplot as plt
```

```python
# α-Beryllium bulk

bs_list = [["monocolor", "bulk hcp beryllium", "4.0_bandstructure/a-Beryllium", "blue"]]

plot_bandstructure("bulk hcp beryllium", bs_list, (-8,8))
plt.savefig("figures/4_bs/4.10_BS_a_bulk.pdf")
plt.savefig("figures_for_publication/S3.15a_bs_hcp.pdf")
```

```python
# α-beryllene

bs_list = [["monocolor", "α-beryllene", "4.0_bandstructure/a-beryllene", "blue"]]

plot_bandstructure("α-beryllene", bs_list, (-8,8))
plt.savefig("figures/4_bs/4.11_BS_a.pdf")
plt.savefig("figures_for_publication/S3.15b_bs_alpha.pdf")
```

```python
# β-beryllene

bs_list = [["monocolor", "β-beryllene", "4.0_bandstructure/b-beryllene", "blue"]]

plot_bandstructure("β-beryllene", bs_list, (-8,8))
plt.savefig("figures/4_bs/4.12_BS_b.pdf")
plt.savefig("figures_for_publication/S3.15d_bs_beta.pdf")
```

```python
# c-Beryllium

bs_list = [["monocolor", "bulk bcc beryllium", "4.0_bandstructure/c0-Beryllium", "blue"]]

plot_bandstructure("bulk bcc beryllium", bs_list, (-8,8))
plt.savefig("figures/4_bs/4.20_BS_c_bulk.pdf")
plt.savefig("figures_for_publication/S3.15c_bs_bcc.pdf")
```

```python
# c-beryllene trilayer

bs_list = [["monocolor", "c-beryllene trilayer", "4.0_bandstructure/c3-beryllene_trilayer", "blue"]]

plot_bandstructure("cubic beryllene trilayer", bs_list, (-8,8))
plt.savefig("figures/4_bs/4.21_BS_c.pdf")
plt.savefig("figures_for_publication/fig3.8_bs_cubic-trilayer-beryllene.pdf")
```

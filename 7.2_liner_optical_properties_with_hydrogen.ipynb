# Linear Optical Properties


```python
# prework

from vmatplot.linear_optical_properties import plot_linear_optical_property
import matplotlib.pyplot as plt

components = [{"xx":"xx-component"}, {"yy":"yy-component"}, {"zz":"zz-component"}]
photon_boundary = (0, 12)
layout = "hor"

# system factors (d_c, (z_max-z_min)*d_c+r_upper+r_lower) r_be = 1.98 r_h = 0.70
factor_bulk = (1, 1)
factor_a = (40, 2*1.98)
factor_b = (40, 0.0473771641975619+2*1.98)
factor_ahh = (40, 0.04361280878130169+2*0.70)
factor_bh = (40, 0.07108860259770651+0.70+1.98)
factor_bhh = (40, 0.0948207854203246+2*0.70)

# systems list
systems = [
           ["α-Beryllene",                "6.0_dielectric_selection/a-Beryllene",     "blue",     None,None,None, factor_a],
           ["2H-terminated α-Beryllene",  "6.0_dielectric_selection/a-Beryllene_hh",  "violet",   None,None,None, factor_ahh],
           ["β-Beryllene",                "6.0_dielectric_selection/b-Beryllene",     "lime",     None,None,None, factor_b],
           ["H-terminated β-Beryllene",   "6.0_dielectric_selection/b-Beryllene_b",   "yellow",   None,None,None, factor_bh],
           ["2H-terminated β-Beryllene",  "6.0_dielectric_selection/b-Beryllene_bb",  "orange",   None,None,None, factor_bhh],
           ]
```

```python
# Absorption coefficient

lim = 0.12
value_boundary = (-0.05*lim, 1.05*lim)
plot_linear_optical_property("Absorption coefficient", systems, "absorption", components, layout, True, "eV", photon_boundary, value_boundary, True)
plt.savefig("figures/7_optical/7.21_absorption_H.pdf")
plt.savefig("figures_for_publication/fig3.19a_abs_H.pdf")
```

```python
# Refractive index

lim = 10
value_boundary = (-0.05*lim, 1.05*lim)
plot_linear_optical_property("Refractive index", systems, "refractive", components, layout, True, "eV", photon_boundary, value_boundary, True)
plt.savefig("figures/7_optical/7.22_refractive_H.pdf")
plt.savefig("figures_for_publication/S3.19a_refractive_H.pdf")
```

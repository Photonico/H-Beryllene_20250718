# Geometry information

```python
# prework

from vmatplot.commons import identify_parameters
import matplotlib.pyplot as plt
```

```python
# a-Beryllium

geometry_data = identify_parameters("2.1_self_consistent/a-Beryllium")
lattice = geometry_data["lattice constant"]
kpoints = geometry_data["kpoints mesh"]
energy = geometry_data["total energy"]
time = geometry_data["elapsed time (sec)"]

print(f"For α-Beryllium bulk: \n k-points mesh: {kpoints}; \n lattice constant: {lattice}; \n total energy: {energy}; \n elapsed time: {time}.")
```

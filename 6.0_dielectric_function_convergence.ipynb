# Dielectric function convergence test


```python
# Prework

from vmatplot.dielectric_function import *
import matplotlib.pyplot as plt

# system factors (d_c, (z_max-z_min)*d_c+2*r_be) r_be = 1.98
factor_bulk = (1, 1)                                        # for α-Beryllium and cubic Beryllium bulk
factor_a = (40, 2*1.98)                                     # for α-Beryllene monolayer
factor_b = (40, 0.0473771641975619+2*1.98)                  # for β-Beryllene monolayer
factor_c = (40, 0.0751622445053417+2*1.98)                  # for cubic Beryllene trilayer

data_boundary = (-60, 60)
```

```python
# α-Beryllium

systems = [["192 nbands",   "6.0_dielectric_function/a-Beryllium_192",  "purple", None,None,None, factor_bulk],
           ["160 nbands",   "6.0_dielectric_function/a-Beryllium_160",  "violet", None,None,None, factor_bulk],
           ["128 nbands",   "6.0_dielectric_function/a-Beryllium_128",  "blue",   None,None,None, factor_bulk],
           ["96 nbands",    "6.0_dielectric_function/a-Beryllium_96",   "lime",   None,None,None, factor_bulk],
           ["64 nbands",    "6.0_dielectric_function/a-Beryllium_64",   "orange", None,None,None, factor_bulk],
           ["32 nbands",    "6.0_dielectric_function/a-Beryllium_32",   "silver", None,None,None, factor_bulk]]

components = [{"xx":"xx-component"},{"yy":"yy-component"},{"zz":"zz-component"}]

plot_dielectric_function("Dielectric function for α-Beryllium", systems, components, "hor", True, "eV", (0,12), data_boundary, True)
plt.savefig("figures/6.10_dielec_a-Beryllium.pdf")
```

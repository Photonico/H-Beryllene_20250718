# Ab initio Molecular Dynamics


```python
# Prework

from vmatplot.aimd import plot_aimd
import matplotlib.pyplot as plt
from vmatplot.output_settings import figure_version, save_figure

figure_version("article")    # "thesis": thesis typography, figures saved as figures/<name>_thesis.pdf
```

```python
# a-beryllene 2H

plot_aimd(
    None, # "double-sided H-terminated α-beryllene",
    "8.0_AIMD/alpha_hh",
    energy_key="F",
    time_step=0.5,
    x_boundary=(0, 5500),
    temperature_boundary=(100, 600),
    energy_color="Blue",
    temperature_color="Cyan",
    energy_line_weight=1,
    temperature_line_weight=1,
)

# fig = plt.gcf()
# fig.set_size_inches(8, 8)
save_figure("aimd_2H-alpha-beryllene")
```

```python
# b-beryllene H

plot_aimd(
    None, # "H-terminated β-beryllene",
    "8.0_AIMD/beta_b",
    energy_key="F",
    time_step=0.5,
    x_boundary=(0, 5500),
    temperature_boundary=(100, 600),
    energy_color="Blue",
    temperature_color="Cyan",
    energy_line_weight=1,
    temperature_line_weight=1,
)

# fig = plt.gcf()
# fig.set_size_inches(8, 8)
save_figure("aimd_1H-beta-beryllene")
```

# Ab initio Molecular Dynamics


```python
# Prework

from vmatplot.aimd import plot_aimd
import matplotlib.pyplot as plt
```

```python
# a-beryllene 2H

plot_aimd(
    None, # "2H-terminated α-beryllene",
    "8.0_AIMD/alpha_hh",
    energy_key="F",
    time_step=1.0,
    x_boundary=(0, 5500),
    temperature_boundary=(100, 600),
    energy_color="Blue",
    temperature_color="Cyan",
    energy_line_weight=1,
    temperature_line_weight=1,
)

plt.savefig("figures/8_aimd/a-beryllene_hh.pdf")
fig = plt.gcf()
fig.set_size_inches(8, 8)
plt.savefig("figures_for_publication/fig3.7a1.pdf")
```

```python
# b-beryllene H

plot_aimd(
    None, # "2H-terminated α-beryllene",
    "8.0_AIMD/beta_b",
    energy_key="F",
    time_step=1.0,
    x_boundary=(0, 5500),
    temperature_boundary=(100, 600),
    energy_color="Blue",
    temperature_color="Cyan",
    energy_line_weight=1,
    temperature_line_weight=1,
)

plt.savefig("figures/8_aimd/b-beryllene_b.pdf")
fig = plt.gcf()
fig.set_size_inches(8, 8)
plt.savefig("figures_for_publication/fig3.7b1.pdf")
```

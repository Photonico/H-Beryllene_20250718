# Phonon dispersion for hexagonal family


```python
# prework

from vmatplot.phonon import plot_phonons
import matplotlib.pyplot as plt

y_range = [-20, 30]
```

```python
# α-Beryllium bulk

matters = [["α-Beryllium", "3.0_phonon_dispersion/a-Beryllium_3", "blue", "solid", 1.5, 1.0, 0]]

plot_phonons("Phonon dispersion for α-Beryllium bulk", matters, y_range, None)
plt.savefig("figures/3.10_phonon_a-Beryllium.pdf")
```

```python
# α-Beryllene

matters = [["α-Beryllene", "3.0_phonon_dispersion/a-Beryllene", "blue", "solid", 1.5, 1.0, 0]]

plot_phonons("Phonon dispersion for α-Beryllene", matters, y_range, None)
plt.savefig("figures/3.11_phonon_a-Beryllene.pdf")
```

## Dynamical stability (PBE, no D3)

Phonopy finite displacements (±0.01 Å) with the relaxation settings (PBE without D3, SIGMA 0.05) in `3.8_phonon_dispersion_pbe`: 5×5×1 supercells with 13×13 k (cubic trilayer 4×4×1 with 16×16 k), bulk hcp 4×4×3 and bulk bcc 4×4×4. The earlier VASP finite-difference runs in `3.0_phonon_dispersion` used PBE+D3, and their "hcp" run was the bcc supercell. The table checks the commensurate q grid (exact for the supercell) and a 60×60 interpolation grid. The cells of α deviate from hexagonal by ~1e-4 Å, so it is analysed with a symmetry tolerance of 1e-3 (P6/mmm); 1e-5 gives Cmmm and an artificial imaginary pocket. Bulk hcp Be is stable everywhere (maximum 20.96 THz, at Γ), bulk bcc Be is harmonically unstable at N (−3.90 THz), as expected for its high-temperature phase.

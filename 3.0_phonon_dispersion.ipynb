```python
# prework

from vmatplot.phonon import plot_phonons
import matplotlib.pyplot as plt

y_range = [0, 30]
```

```python
# α-Beryllium bulk

matters = [["α-Beryllium", "3.0_phonon_dispersion/a-Beryllium_3", "blue", "solid", 1.5, 1.0, 0]]

plot_phonons("Phonon dispersion for bulk hcp beryllium", matters, y_range, None)
plt.savefig("figures/3_phonon/3.10_phonon_a-Beryllium.pdf")
plt.savefig("figures_for_publication/S3.13a_phonon_hcp.pdf")
```

# Convergence testing for cubic family

Energy versus parameters


```python
# prework

from vmatplot.convergence import summarize_parameters, plot_energy_parameters
import matplotlib.pyplot as plt

encut_boundary = (300, None)
kpoints_boundary = (12**2, None)
```

```python
# energy versus k-points for c-beryllene bulk

summarize_parameters("1.0_convergence/c0-Beryllium")
t_kpoints = ["bulk bcc beryllium", "1.0_convergence/c0-Beryllium/energy_parameters.dat", kpoints_boundary, "blue"]
kpoints_list = [t_kpoints]

plot_energy_parameters("Energy versus k-points for bulk bcc beryllium", "kpoints", kpoints_list)
plt.savefig("figures/1_convergence/1.5_conv_c0.pdf")
plt.savefig("figures_for_publication/S3.2c_conv_bcc.pdf")
```

```python
# energy versus k-points for c-beryllene monolayer

summarize_parameters("1.0_convergence/c1-beryllene_monolayer")
c_kpoints = ["c-beryllene", "1.0_convergence/c1-beryllene_monolayer/energy_parameters.dat", kpoints_boundary, "blue"]
kpoints_list = [c_kpoints]

plot_energy_parameters("Energy versus k-points for c-beryllene monolayer", "kpoints", kpoints_list)
plt.savefig("figures/1_convergence/1.6_conv_c1.pdf")
```

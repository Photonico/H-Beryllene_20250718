# Convergence testing for cubic family

Energy versus parameters


```python
# prework

from vmatplot.convergence import summarize_parameters, plot_energy_parameters
import matplotlib.pyplot as plt

encut_boundary = (300, None)
kpoints_boundary = (12**2, None)
```

```python
# energy versus k-points for c-Beryllene bulk

summarize_parameters("1.0_convergence/c0-Beryllene_bulk")
t_kpoints = ["c-Beryllene bulk", "1.0_convergence/c0-Beryllene_bulk/energy_parameters.dat", kpoints_boundary, "blue"]
kpoints_list = [t_kpoints]

plot_energy_parameters("Energy versus k-points for c-Beryllene bulk", "kpoints", kpoints_list)
plt.savefig("figures/1.conv_c0-Beryllene.pdf")
```

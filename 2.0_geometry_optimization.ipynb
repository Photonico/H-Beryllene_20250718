# Geometry optimization

Energy versus parameters

* VASP parameters:

`ISIF = 3`

```python
# prework

from vmatplot.convergence import summarize_parameters, plot_energy_parameters, summarize_cohesive_energy, plot_cohesive_energy_parameters
from vmatplot.commons import identify_parameters

import matplotlib.pyplot as plt
```

```python
# energy versus lattice constant for g-Beryllene

summarize_parameters("1.0_convergence/g-Beryllene_lattice")
g_lattice = ["for g-Beryllene", "1.0_convergence/g-Beryllene_lattice/energy_parameters.dat", None, None, "blue"]
lattice_list = [g_lattice]

plot_energy_parameters("Energy versus lattice constant for g-Beryllene", "lattice", lattice_list )
plt.savefig("figures/1.21_lattice_g.pdf")
```

```python
# cohesive energy versus lattice constant for g-Beryllene

summarize_cohesive_energy("1.0_convergence/g-Beryllene_lattice", "1.0_convergence/single_Beryllium")
g_lattice = ["for g-Beryllene", "1.0_convergence/g-Beryllene_lattice/cohesive_energy.dat", None, None, "blue"]
lattice_list = [g_lattice]

plot_cohesive_energy_parameters("Cohesive energy versus lattice constant for g-Beryllene", "lattice", lattice_list)
plt.savefig("figures/1.22_lattice_g.pdf")
```

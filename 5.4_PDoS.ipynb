# Projected Density of State


```python
# prework

from vmatplot.pdos import plot_single_pdos
import matplotlib.pyplot as plt
```

```python
# α-Beryllium bulk

be_index = [1]
current_dict = "5.0_PDoS/a-Beryllium"

systems = [
           ["Total DOS", current_dict, [], "total_dos", "silver", "solid", 2.0, 1.0],
           ["Total PDoS", current_dict, [], "total", "grey", "solid", 2.0, 1.0],
           ["$p$", current_dict, be_index, "p", "blue", "solid", 2.0, 1.0],
           ["$s$", current_dict, be_index, "s", "blue", "dashed", 2.0, 1.0]
           ]

plot_single_pdos("PDoS for α-Beryllium bulk", systems, 4, 1.2)
plt.savefig("figures/5_pdos/5.10_a-Beryllium.pdf")
```
